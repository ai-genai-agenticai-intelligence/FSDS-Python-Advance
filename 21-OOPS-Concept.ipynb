     01-Task: Create a Student class that counts how many students exist, gives  a   grade (A ≥ 75, B ≥ 50, C ≥ 35, otherwise Fail), and has a helper that checks pass or fail.?

```python
class Student:
    school = "ABC School"          # class variable
    count = 0                      # class variable (static counter)

    def __init__(self, name, marks):
        self.name = name           # instance variables
        self.marks = marks
        Student.count += 1

    def grade(self):               # instance method
        if self.marks >= 75:
            return "A"
        elif self.marks >= 50:
            return "B"
        elif self.marks >= 35:
            return "C"
        return "Fail"

    @classmethod
    def total_students(cls):       # class method
        return cls.count

    @staticmethod
    def is_pass(marks):            # static method
        return marks >= 35


s1 = Student("Ravi", 82)
s2 = Student("Asha", 48)
s3 = Student("Kiran", 30)

for s in (s1, s2, s3):
    print(s.name, s.grade())

print(Student.total_students())
print(Student.is_pass(34))
```

```python
class A:
    def feature1(self):
        print('feature is hii')

    def feature2(self):
        print('feature is hello')


class B(A):
    def feature3(self):
        print('feature is Welcome')

    def feature4(self):
        print('feature is good morning')    # fixed typo "feaeture"


class C(B):                                  # B already inherits A
    def feature5(self):
        print('feature is good evening')

    def feature6(self):
        print('feature is good night')


a1 = A()
a1.feature1()
a1.feature2()

a2 = B()
a2.feature1()      # inherited from A
a2.feature2()      # inherited from A
a2.feature3()
a2.feature4()

a3 = C()           # () creates the object
a3.feature1()      # from A
a3.feature3()      # from B
a3.feature5()      # its own
```

      4-Hybrid Inheritence
      ---combine all(Multiple & multilevel inheritence)

```python
class A:
    def showA(self):
        print('Class A')


class B:
    def showB(self):
        print('Class B')


class C(A, B):
    def showC(self):
        print('Class C')


class D(C):                  # C already brings in A and B
    def showD(self):
        print('Class D')


d1 = D()                     # () creates objects
c1 = C()

c1.showC()
d1.showD()

# D inherits everything from C, A and B
d1.showA()
d1.showB()
d1.showC()
```

#  7-Constructer -inheritence
        ---A constructer is a special method
        --it is autometically called when an object is created
        --In inheritence ,constructer of parent and child class can work togather using super()
        sup().__init__()
        --sup() it will First called parent class constructer

```python
class A():

    def __init__(self):
        print('This is a constructer')
    def feature1(self):
        print('This is a feature1')
    def feature2(self):
        print('This is a feature2')

        
class B(A):
    def __init__(self):
        super().__init__()
        print('This is a constructer of class B')
    def feature3(self):
        print('This is a feature 3')
    def feature4(self):
        print('This is a feature4')

b1 =B()
b1.feature1()
b1.feature3()
b1.feature2()
        
```

```python
class Parent:
    def __init__(self, name):
        self.name = name


class Child(Parent):                    # inherit from Parent
    def __init__(self, name, age):
        super().__init__(name)          # runs Parent.__init__ -> sets self.name
        self.age = age                  # Child's own variable


obj = Child('Abhishek', 22)

print(obj.name)
print(obj.age)
```

     # Duck Typing
      --Duck typing behaviour--class name always Different but must should method name same

```python
class Dog:
    def speak(self):
        print("Woof")
        
class Cat:
    def speak(self):
        print("Meow")

class Robot:
    def speak(self):
        print("Beep boop")

def display(obj):          # obj can be ANY object that has speak()
    obj.speak()


d1 = Dog()
c1 = Cat()
r1 = Robot()

display(d1)
display(c1)
display(r1)
```

```python
class Duck:                    # different class
    def speak(self):           # same method name
        print("Quack")

class Dog:                     # different class
    def speak(self):           # same method name
        print("Woof")

class Robot:                   # different class
    def speak(self):           # same method name
        print("Beep boop")


def make_it_speak(thing):
    thing.speak()              # only needs the method name "speak"


make_it_speak(Duck())
make_it_speak(Dog())
make_it_speak(Robot())
```

# 12-Multitherding
     ---multiple task perform at a same time 
     --- it is save my time
# Thread
    ---it is save my time 
    ---we multiple task at a time
    ---we run get Simouttously

```python
class Hello:
    def run(self):
        for i in range(5):
            print("hello")


class Hii:
    def run(self):
        for i in range(5):
            print("hii")


t1 = Hello()
t1.run()

t2 = Hii()    # Create object
t2.run()
```

```python
from threading import *

class Hello(Thread):
    def run(self):
        for i in range(5):
            print("hello")

class Hii(Thread):
    def run(self):
        for i in range(5):
            print("hi")

t1 = Hello()
t2 = Hii()

t1.start()    # Start Thread 1
t2.start()    # Start Thread 2

t1.join()     # Wait for Thread 1
t2.join()     # Wait for Thread 2

print("Main Thread Ended")
```
