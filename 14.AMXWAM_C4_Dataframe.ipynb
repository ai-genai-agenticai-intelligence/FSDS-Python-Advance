# we analyze python dataframe or dataset

```python
import matplotlib.pyplot as plt
import seaborn as sns

%matplotlib inline
plt.rcParams['figure.figsize'] = 6,2

import warnings
warnings.filterwarnings('ignore')
```

```python
df.head()
```

```python
df['InternetUsers']
```

```python
# Distribution
vis1 = sns.distplot(df['InternetUsers'])
plt.show(vis1)
```

```python
vis2 = sns.displot(df['InternetUsers'])
plt.show()
```

```python
vis3 = sns.distplot(df['InternetUsers'],bins=15,color='blue')
vis3
```

```python
# Box PLot
vis4 = sns.boxplot(data = df,x='IncomeGroup',y='BirthRate')
vis4
```

```python
vis5 = sns.lmplot(data = df,x ='InternetUsers',y='BirthRate')
plt.show()
```

```python
vis6 = sns.lmplot(data = df,x = 'InternetUsers', y = 'BirthRate', fit_reg = True ) #lm - linear model
plt.show()
```

```python
vis8 = sns.lmplot(data = df,x ='InternetUsers',y = 'BirthRate',fit_reg = False,hue='IncomeGroup') #hue-Parameter for color
plt.show()
```

```python
vis5 =sns.lmplot(data=df,x='InternetUsers',y='BirthRate',fit_reg=True,hue = 'IncomeGroup')
vis5
```
