# Python Function
      1-Userdefined function
      2-inbuiled function
      
      3-Arguments
          -Positional Argument      →Arguments are passed according to their position.
          -Key Argument             →Arguments are passed using parameter names.
          -default Argument         →A default value is assigned to a parameter.
          -Variable length argument →Used when the number of arguments is unknown.
          
       -It is mainly 2 types
           -*args        →*args collects multiple positional arguments.Stored as a tuple.
           -**kwargs     → **kwargs collects multiple keyword arguments.Stored as a dictionary.
           
     4-lambda
         -filter()       →Used to filter elements based on a condition.
         -map()          →Used to transform each element
         -reduce()       →Used to reduce multiple values to a single value
         
     5-Error
         -logical error  →Program runs but gives wrong output.
         -compile error  →Error occurs due to invalid syntax before program execution Example: SyntaxError
         -runtime error  →Error occurs while the program is running. Example: ZeroDivisionError, NameError, TypeError
                             
         
     6-Error2
         -try:           →risky code
         -except         → handles error
         -Exception as e → stores error message
         -finally        → executes always

```python
def count(lst):

    even = 0
    odd = 0

    for i in lst:
        if i%2==0:
            even+=1
        else:
            odd+=1
    return even,odd

lst = [1,2,3,4,5,6,7,8,9]
even,odd =count(lst)

print(even)
print(odd)
    
```
